# 03 - Auto-ML Model Selection

## Komparasi Algoritma SOTA
Kita akan melatih Linear Regression, Random Forest, XGBoost, LightGBM, dan CatBoost untuk mencari model terbaik dalam memprediksi RVI Score.

```
import sys
sys.path.append('../src')
from etl import DataLoader
from preprocessing import DataPreprocessor
from feature_engineering import FeatureEngineer
from models import ModelTrainer

df = DataLoader('../data/raw/master_dataset.csv').load_raw_data()
clean_df = DataPreprocessor(df).handle_missing_values()
final_df = FeatureEngineer(clean_df).execute_all()
```

```
trainer = ModelTrainer(final_df)
results_df = trainer.train_and_evaluate()
display(results_df)

trainer.save_best_model('../models/best_model.pkl')
```

## Insight Modeling
Model dengan RMSE terendah dan R2 tertinggi (biasanya Tree-based Ensemble seperti XGBoost atau CatBoost) dipilih karena kemampuannya menangani non-linearitas dalam data iklim dan sosio-ekonomi.
