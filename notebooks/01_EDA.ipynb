# 01 - Exploratory Data Analysis (EDA)

## 1. Dataset Overview
Mengapa ini penting? Sebelum melakukan inferensi model, integritas data (Missing Values, Duplikasi, Outlier) harus dipastikan untuk mencegah *Garbage-In-Garbage-Out*.

```
import pandas as pd
import sys
sys.path.append('../src')
from etl import DataLoader

df = DataLoader('../data/raw/master_dataset.csv').load_raw_data()
display(df.head())
display(df.info())
```

## 2. Missing Value Analysis
Polanya harus dianalisis apakah *Missing Completely at Random* (MCAR) atau memiliki pola khusus.

```
import missingno as msno
import matplotlib.pyplot as plt

msno.matrix(df, figsize=(10, 5))
plt.title('Missing Value Matrix')
plt.show()
```

## 3. Distribution & Outlier Analysis
Distribusi target dan fitur sangat krusial untuk asumsi regresi linear dan robustnes model *tree-based*.

```
from visualization import DataVisualizer
import seaborn as sns

DataVisualizer.plot_distribution(df['produksi_padi'], 'Distribusi Produksi Padi')
plt.show()
```

```
DataVisualizer.plot_boxplots(df, ['curah_hujan', 'suhu'], 'Outlier Deteksi Iklim')
plt.show()
```

## 4. Feature Correlation Ranking
Menemukan multikolinearitas dan korelasi linier dengan target.

```
DataVisualizer.plot_correlation_matrix(df)
plt.show()
```

## 5. Insights
- **Iklim**: Terdapat beberapa outlier pada curah hujan yang mewakili fenomena ekstrim (El Nino/La Nina).
- **Korelasi**: Suhu dan curah hujan memiliki korelasi yang signifikan terhadap volatilitas produksi padi.
